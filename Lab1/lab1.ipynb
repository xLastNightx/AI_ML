```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler

print("pandas:", pd.__version__)
```

```python
df = pd.read_csv("titanic.csv")
df.head()
```

```python
print("Размер датасета:", df.shape)
print("\nПервые 40 строк:")
df.head(40)
```

```python
#подсчет значений
numeric_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
categorical_cols = ['HomePlanet', 'Destination', 'Cabin']
bool_cols = ['CryoSleep', 'VIP']

medians = {col: df[col].median() for col in numeric_cols}
# medians = {}
# for col in numeric_cols:
#     medians[col] = df[col].median()
# print(medians)
modes_cat = {col: df[col].mode()[0] for col in categorical_cols}
modes_bool = {col: df[col].mode()[0] for col in bool_cols}

print("Медианы:", medians)
print("Моды категориальных:", modes_cat)
print("Моды булевых:", modes_bool)
```

```python
# заполнение
for col in numeric_cols:
    df[col] = df[col].fillna(medians[col])

for col in categorical_cols:
    df[col] = df[col].fillna(modes_cat[col])

for col in bool_cols:
    df[col] = df[col].fillna(modes_bool[col])
```

```python
for col in bool_cols:
    df[col] = df[col].astype(bool)

df[bool_cols].dtypes
```

```python
df.isnull().sum()
```

```python
# остались имена не заполненные некоторые, чтобы полностью выполнить 4 пункт заполню их модой(сейчас вычисление моды)
name_mode = df['Name'].mode()[0]
print("Мода Name:", name_mode)
```

```python
# заполнение
df['Name'] = df['Name'].fillna(name_mode)
```

```python
df.isnull().sum()
```

```python
#нормализация MinMaxScaler приводит все значения к диапазону [0, 1] по формуле: X' = (X - X_min) / (X_max - X_min)
from sklearn.preprocessing import MinMaxScaler

numeric_to_scale = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']

scaler = MinMaxScaler()
df[numeric_to_scale] = scaler.fit_transform(df[numeric_to_scale])

df[numeric_to_scale].describe()
```

```python
#Преобразование категориальных
df[['Deck', 'CabinNum', 'Side']] = df['Cabin'].str.split('/', expand=True)
df = df.drop(['Cabin', 'CabinNum'], axis='columns')
```

```python
#удаляем лишнее
df = df.drop(['Name', 'PassengerId'], axis='columns')
```

```python
#One-Hot Encoding
df = pd.get_dummies(df, columns=['HomePlanet', 'Destination', 'Deck', 'Side'], drop_first=True)
df.head()
```

```python
#check
print("Размер:", df.shape)
print()
print(df.dtypes)
```

```python
#bool к int
for col in df.columns:
    if df[col].dtype == bool:
        df[col] = df[col].astype(int)

df.dtypes
```

```python
print("Размер:", df.shape)
print()
print(df.dtypes)
print()
print("Пропусков:", df.isnull().sum().sum())
```

```python
#сохранение
df.to_csv("processed_spaceship_titanic.csv", index=False)
print("Сохранено в processed_spaceship_titanic.csv")
```
